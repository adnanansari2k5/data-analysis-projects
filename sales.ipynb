```python
import pandas as pd

data = pd.read_csv("sales_data.csv")

print(data.head())
```

```python
print("Total Sales:", data["Sales"].sum())
print("Total Profit:", data["Profit"].sum())
```

```python
print(data.groupby("Region")["Sales"].sum())
```

```python
import matplotlib.pyplot as plt

region_sales = data.groupby("Region")["Sales"].sum()

plt.figure(figsize=(8,5))
region_sales.plot(kind='bar')

plt.title("Sales by Region", fontsize=14)
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.savefig("sales_region.png")
plt.show()
```

```python
product_profit = data.groupby("Product")["Profit"].sum()
print(product_profit)
```

```python
import matplotlib.pyplot as plt

product_profit = data.groupby("Product")["Profit"].sum()

plt.figure(figsize=(6,6))
product_profit.plot(kind='pie', autopct='%1.1f%%')

plt.title("Profit Distribution by Product")

plt.ylabel("")   # y-axis hata deta hai (clean look)
plt.tight_layout()

plt.savefig("profit_pie.png")
plt.show()
```
